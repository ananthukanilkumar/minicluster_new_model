```python
import numpy as np
import matplotlib.pyplot as plt
data = np.load('minicluster_results100_p1e7.npz')

print(data.files)  # lists the saved array names

SD_1 = data['SD_1']
SD_2 = data['SD_2']
SD_3 = data['SD_3']
minicluster_mass = data['minicluster_mass']
minicluster_radius = data['minicluster_radius']
minicluster_start_pos = data['minicluster_start_pos']

for k in data.files:
    print(k, data[k].shape, data[k].dtype)
```

```python
plt.plot(SD_1,label='SD_1')
plt.yscale('log')
```

```python
plt.hist(SD_1, bins=np.logspace(-8,1,50),color='red',edgecolor='k',lw=1.2,alpha=0.5)
plt.hist(SD_2, bins=np.logspace(-8,1,50),color='green',edgecolor='k',lw=1.2,alpha=0.5)
plt.hist(SD_3, bins=np.logspace(-8,1,50),color='blue',edgecolor='k',lw=1.2,alpha=0.5)

plt.xscale('log')
```

```python
plt.hist([SD_1, SD_2, SD_3], bins=np.logspace(-8,1,50), histtype='step', linewidth=1.5)
plt.xscale('log')
```
